## self

- زمانی که در کلاس میخواهیم پارامتری تعریف کنیم نیاز است ان را توسط سلف به عنوان شی وارد متد کنیم.

```python
class Student:
    def __init__(self, name, grade):   # سازنده کلاس
        self.name = name
        self.grade = grade

    def introduce(self):
        print(f"My name is {self.name} and my grade is {self.grade}")

# self.name و self.grade ویژگی‌های هر شیء دانشجو هستند.
# self در واقع اشاره به همان شیء دارد.


# حالا وقتی ما شی را میسازیم
student1 = Student("Ali", 18)

# یعنی در عمل
Student.__init__(student1, "Ali", 18)


```

## چرا از سلف استفاده میکنیم

- چون در OOP ممکن است چند نمونه از یک کلاس ساخته شود، و هرکدام باید داده‌ی خودش را نگه دارد.

```python
# EX02

class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"{self.owner}'s new balance = {self.balance}")

acc1 = BankAccount("Ali", 1000)
acc2 = BankAccount("Sara", 3000)

acc1.deposit(500)
acc2.deposit(200)

```

## Encapsulation (کپسول سازی)

- محافظت از داده های داخل کلاس : یعنی ما اجازه  نمیدهیم هر بخشی از برنامه مستقیما داده های حساس را تغییر دهد .




```python
class BankAccount:

    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount

    def withdraw(self, amount):

        if amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Not enough balance")

    def show_balance(self):
        print(self.__balance)


acc = BankAccount("Ali", 1000)

acc.show_balance()
acc.deposit(500)
acc.show_balance()
acc.withdraw(200)
acc.show_balance()
```
