## Inheritance(وراثت)

- یعنی : ارث بردن یک کلاس از کلاس دیگر

    - -> یک کلاس (کلاس فرزند) می تواند ویژگی ها و ممتدهای کلاس دیگر (کلاس والد) را داشته باشد.

    - -> سپس میتواند اضافه کند یا رفتار ها را عوض کند .

```python
class Animal():
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(f"{self.name} makes a sound")

class Dog(Animal):
    def speak(self):
        print(f"{self.name} barks")

d = Dog("Rex")
d.speak()
```

- Dog(Animal) یعنی Dog از Animal ارث برده

- چون Dog متد speak() را دوباره تعریف کرده، رفتار جدید جایگزین می‌شود. این Override است.

## __init__ در وراثت و نقش super()

- اگر کلاس فرزند اینیت مخصوص خودش را نداشته باشد ، از والد استفاده می کند ، اما اگر فرزند اینیت داشته باشد باید مقادیر پایه والد هم تنظیم شود 

```python
class Animal:
    def __init__(self, name):
        self.name = name

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)   # صدا زدن سازنده والد
        self.breed = breed

d = Dog("Rex", "German Shepherd")
print(d.name)   # Rex
print(d.breed)  # German Shepherd
```

```python
# EX 01

class Payment():
    def __init__(self, amount):
        self.amount = amount

    def pay(self):
        raise NotImplementedError("Subclasses must implement pay()") # یعنی این متد نباید در کلاس پایه اجرا شود 
        

```

### نکته :

### raise :

- در پایتون برای ایجاد عمدی یک استثنا استفاده می شود .

### NotImplementedError:

- یک نوع خاص از خطا است که معمولاً زمانی استفاده می‌شود که:

    - یک کلاس پایه (Base Class) متدی را تعریف کرده
    - ولی اجرای واقعی آن باید توسط زیرکلاس‌ها (Subclasses) انجام شود.

- این خطا نشانه می‌دهد که متد در کلاس پدر عمداً اجرا نشده و باید در کلاس فرزند override شود.

```python
class CreditCarsPayment(Payment):  #پرداخت با کارت اعتباری
    def __init__(self, amount, card_number):
        super().__init__(amount=amount)

        self.card_number = card_number

    def pay(self):
        print(f"Paid {self.amount} using credit card {self.card_number}")


class CashPayment(Payment):         #پرداخت نقدی
    def pay(self):
        print(f"Paid {self.amount} in cash")

p1 = CreditCarsPayment(500, "1234-5678")
p1.pay()

p2 = CashPayment(200)
p2.pay()
```

## polymorphism (چند ریختگی)

- یعنی چند کلاس مختلف یک متد مشترک دارند ، اما رفتار متفاوتی انجام می دهند .

#### مثال ساده :

- فرض کنید سه کلاس با عنوان سه حیوان داریم ، همه کلاس ها یک متد مشترک دارن به نام سپیک اما صدایشان متفاوت است .

```python
# EX 01

class Dog():
    def speak(self):
        print("Dog barks")

class Cow:
    def speak(self):
        print("Cow moos")

class Cat:
    def speak(self):
        print("Cat meows")

animals = [Dog(), Cow(), Cat()]

for animal in animals:
    animal.speak()
```

```python
class Animal:
    def speak(self):
        print("Animal makes a sound")

# کلاس فرزند

class Dog(Animal):
    def speak(self):
        print("Dog barks")

class Cat(Animal):
    def speak(self):
        print("Cat meows")

animals = [Dog(), Cat()]

for animal in animals:
    animal.speak()
```
