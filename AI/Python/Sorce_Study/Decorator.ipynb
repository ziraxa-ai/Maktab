## Decorators (دکوراتور)

#### - توابع در پایتون به عنوان شی

    - -> می توانیم در پایتون ، یک متغیر را با مقدار یک تابع مقدار دهی کنیم 

#### - توابع مرتبه بالاتر
    
    - -> تابعی که تابع دیگر را بگیرد یا برگرداند 

#### - سینتکس دکوراتور (@)

    - -> پایتون برای خوانایی بیشتر این سینتکس را درنظر گرته 



    

```python
def my_decorator(func):
    def wrapper():
        print("Before function")
        func()
        print("after")
    return wrapper

def say_hello():
    print("hello")

say_hello_ = my_decorator(say_hello)
say_hello_()
```

```python
# سینکتکس دکوراتور

@my_decorator
def say_bay():
    print("bay")
```

#### حالا اگر سنتکسمون پارامترداشت 



```python
def repeat(n):

    def decorator(func):

        def wrapper(*args, **kwargs):

            for i in range(n):
                func(*args, **kwargs)

        return wrapper
    
    return decorator

@repeat(3)
def hello():
    print("hi")
hello()
```

#### دکوراتور کلاس

    - -> دکوراتورها فقط برای تابع نیستند ؛ برای کلاس هم استفاده میشوند .
    

```python
def add_method(cls):

    def new_method(self):
        print("New method")

    cls.new_method = new_method
    return cls


@add_method
class MyClass:
    pass
```

```python
# نکته ما میتوانیم داخل تابعمون یک تابع دیگر داشته باشیم 
def calculated(a, b , what_to_do):
    def add(a, b):
        return a + b
    
    def zarb(a, b):
        return a * b
    
    if what_to_do == "add":
        return add(a, b)
    elif what_to_do == "zarb":
        return zarb(a, b)
    
result = calculated(3, 4, "zarb")
print(result)
    
```

```python
# گاهی اوقات هم ما برای جلوگیری از تکرار بی مورد نیاز داریم یک فانکشن را به گونه ای بنویسیم که یک فانکشن ورودی بگیرد و درصورت درست بودن شرط پیش برود 

def run_on_zoj(f):
    import datetime 
    now = datetime.datetime.now()
    minute = now.minute
    if minute % 2 == 0:
        f()
    else:
        print("hisss")

def say_hello():
    print("salam")

def say_bye():
    print("bye")

run_on_zoj(say_hello)
run_on_zoj(say_bye)

# تا اینجا با مفهوم آنشا شدیم حالا فانکشن دکورتور را امتتتحان میکنیم




```

```python
# EX 01

import time

def timer_decorator(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"{func.__name__} executed in {end - start :.4f} seconds")
        return result
    return wrapper

@timer_decorator
def load_data():
    time.sleep(2)

@timer_decorator
def analyze_data():
    time.sleep(3)  # شبیه‌سازی تحلیل داده

load_data()
analyze_data()
```

```python
# EX 02 
def log_decorator(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print(f"Executing {func.__name__} with args={args} kwargs={kwargs}")
        result = func(*args, **kwargs)
        print(f"{func.__name__} returned {result}")
        return result
    return wrapper

@log_decorator
def compute_mean(data):
    return sum(data) / len(data)

# نمونه داده:
compute_mean([1, 2, 3, 4, 5])
```
