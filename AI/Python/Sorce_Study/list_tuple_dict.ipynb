## Data structures

### list

- لیست ترتیب دارد
- تمامی امکانات اسلاسینگ و ایندکسی که بلدیم اینجا کاربرد دارد
- می تواند هر چیزی درونش داشته باشد

```python
my_list = [0,1,2,3,4]
namse = ["Taha" , "aty", "saadat", "sadaf"]

print(namse[1:3])
print(namse[-1])
print(namse[0:3:2])
print(len(namse))
print(len(namse[-1]))
print(my_list* 2)
```

### dict

 - key : Value
 - 

```python
price = {"apple" : 100, "bana": 2, "khiar": 1000}
print(price["apple"])

all = {
    "today" : {
        "apple" : 100,
        "bana": 2,
        "khiar": 1000
        },
    "after" : {
        "apple" : 530,
        "bana": 20,
        "khiar": 10
        }
    }

all_list = {
    "list_1" : [1,2,3,4,5,6],
    "list_2" : [6,5,4,3,2,1]
}

all_list_2 = {
    "list_1" : ["taha", "arman", "arash"],
    "list_2" : ["saadat", "messi", "ajaly"]
}

# slicing
print(all_list["list_2"][1:5:2])
print(all_list_2["list_2"][1].upper()[2])
```

### tuple 

- اسلایسینگ و دسترسی کاملا همانند 
- تاپل ابجکت هستش وامکانات تغییرات ندارد (imutable)



```python
tuple_test = (1, "taha")

name, score = tuple_test
print(name)
print(score)
```

### set(مجموعه)

- ابتدا باید تعریف شود 
- ترتیب ندارد 
- جنس تکراری نمی پذیرد 


```python
set_test = set()
set_test.add(1)
print(set_test)
set_test.add(4)
set_test.add(1)
print(set_test)


heyvoons = set(["dog", "gorg", "sag", "fok", "sag", "sag"])
print(heyvoons)
```
