## انواع ارور ها 

- -> یک نوع اروری داریم که که منطق برنامه نویسی است .

- -> یک نوع ارور هم به علت نبودن یا اشتباه نوشتن یک تکست به وجود می آید .

- -> یک موقعی هست که ارور ما برای زمانیه هست که مثلا میخواهیم یک فایل را باز کنیم و نیست یا خالی است .

 ما نمی توانیم در همه جا ارور ها رو پیش بینی کنیم پس از متد های استفاده میکنیم 
 
 try :  <=> سعی کن این کار رو انجام بدی 



 except:  <=> اگر نشد این رو انجام بده

 این قابلیت افزودن و یا قراردادن شرط را هم دارد 
 
### در مثال ببینید


```python
try:
    prit("this is test .")

except:
    print("test ")



# برنامه ای داریم برای تقسیم اعداد و احتمال دارد به مشکل اروری مثلا صفر یا اشتباه در دادن ورودی بر بخوریم 

def divis(a, b):
    try:
        return a / b


    except ZeroDivisionError:
        print("b can not be 0")

    except Exception as e :   # عنوان خطا رو در e بریز
        print(f"Error in returnung, Error : {e}")
        return None

print(divis(3, 0))
print("--------------------")
print(divis("3", 0))
```

ما میتوانیم ارور هامون رو پیش بینی کنیم و براشون شرط بزاریم تا کاری جداگانه انجام دهند  یا در صورت برخورد با هر اروری چه رفتاری نشان دهند .


- -> else: <=> اگر به اروری برنخوردی.

- -> finally: <=> در نهایت این کار ها رو هم انجام بده .

خوبه که ما در ترای فقط کار هایی رو در نظر بگیریم که به مشکل بر بخوره 



```python
def divis(a, b):
    try:
        r = a / b

    except Exception as e:
        print(f"had error {e}")

    else:   # در صورتی ران میشود که ما به اروری نخوریم 
        print("else")

    finally:   # همیشه ران میشه
        print("finally")
```
