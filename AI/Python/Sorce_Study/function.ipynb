```python
# مثال تکمیل تر 

def hello(name):
    '''
    This function sey hello to the name,
    to the length of name.                  
    '''                                     # ترجیحا توضیحات در توابعتون بگذارید
    for i in range(len(name)):
        print(f"oh hello {name}")

hello(name="taha")        
```

## انتخاب پیش فرض

- ما می توانیم برای ایتم هایی که در فانکشن درست می کنیم یک پیش فرض و یک جنسیت اماده بکنیم به شکل زیر 



```python
def print_times(name, n = 1):
    for i in range(n):
        print(name)

print_times("taha")
print_times("--------------")
print_times("taha", n = 5)
```

## *args

- در پرانتز تابع تعریف میشود و یک تاپل به داخل تابع میبرد بجای تعریف یکی یکی 

## **kwargs

- در این فانکشن یک یکشنری کامل درست میکند که میتوانیم باهاش کار کنیم 

```python
def  zarb(*args):
    print(f"args :{args}")
    for i in args:
        print(i)

print(zarb(2,4,6,7,8,3,2))
```

```python
def contact(**kwargs):
    print(kwargs)

contact(name= "taha", family = "saadat" , age = 21)
```

## Anonymous Function

- lambda --> زمانی که می خواهیم یک فانکشن نامرئی طراحی بکنیم به این شکل خواهد بود 

- map --> میگه من یک تابع رو روی اعضای یک لیست ران میکنم

- filter --> من اعضای یک لیست رو فیلتر میکنم


```python
def tavan_2(x):
    return x ** 2

list_ = [2,3,4,6,5]


# روش سنتی
for i in range(len(list_)):
    list_[i] = tavan_2(x=list_[i])
print(f"list sonaty :{list_}")

list_ = [2,3,4,6,5]
# map
# فرایند بالا به کمک مپ 
tavan_2_list = map(tavan_2, list_)
print(f"list map  :{list(tavan_2_list)}")

# lambda
# به جای روش بالا به گونه زیر تابع را تعریف میکنیم
tavan_2_list = map(lambda x : x ** 2 , list_)
print(f"list lambda _map  :{list(tavan_2_list)}")

# filter
# یک تابع مشخص میکنیم و ازفیلتر برای اجرای آن روی یک لیست استفاده میکنیم

def is_ashari(x):
    return x != int(x)

list_ = [5, 6.5, 8, 9.1, 8, 9, 4, 2.9]
print(list(filter(is_ashari, list_)))

# حالا به کمک لامبدا فیلتر میکنیم برای آن هایی که اعشار نیست
print(list(filter(lambda x : x == int(x), list_)))
```
