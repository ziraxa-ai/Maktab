## Abstract Class (کلاسی که فقط قابل را تعریف میکند)

- می گوید کلاس های فرزند باید چه متد هایی داشته باشند اما خودش آن متد ها را ککامل  پیاده سازی نمیکند .


```python
# برای ساخت این کلاس ها باید کتابخانه ای بی سی را ایمپورت کنیم 

from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def speak(self):
        pass

class Dog(Animal):
    def speak(self):
        print("Dog barks")

class Cat(Animal):
    def speak(self):
        print("Cat meows")

animals = [Dog(), Cat()]

for a in animals:
    a.speak()

    
```

```python
# __repr__

class Car:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed

    def __str__(self):
        return f"Car({self.brand}, {self.speed})"

    def __repr__(self):
        return f"Car(brand='{self.brand}', speed={self.speed})"
    
C = Car("Benz", 220)
repr(C)
```

```python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def __add__(self, other):
        return BankAccount(self.balance + other.balance)

    def __str__(self):
        return f"Balance: {self.balance}"

a1 = BankAccount(100)
a2 = BankAccount(200)
a3 = a1 + a2
print(a3)
```

```python
# __eq__, __lt__

class Car:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed

    def __eq__(self, other):
        return self.speed == other.speed

    def __lt__(self, other):
        return self.speed < other.speed

c1 = Car("BMW", 220)
c2 = Car("Audi", 200)

print(c1 == c2)  # False
print(c1 < c2)   # False
```

```python
# __enter__ و __exit__
# برای استفاده در ساختار with
# (مدیریت منابع مثل فایل‌ها، پایگاه‌داده و غیره)
class FileManager:
    def __init__(self, filename):
        self.filename = filename

    def __enter__(self):
        self.file = open(self.filename, 'w')
        return self.file

    def __exit__(self, exc_type, exc_val, exc_tb):
        self.file.close()

with FileManager("data.txt") as f:
    f.write("Hello World")
```

##  Composition (ترکیب) و Aggregation (تجمیع)

- اما در دنیای واقعی و برنامه‌نویسی حرفه‌ای، بیشتر با رابطه‌ی “دارد” (Has-a) سر وکارداریم.

- مثلاً: Car یک Engine (موتور) دارد. (نه اینکه ماشین خودش یک نوع موتور باشد!)به این مفهوم Composition (ترکیب) می‌گوییم.

### Is-a <=> هست

### Has-a <=> دارد



```python
# مثال برای ساخت یک کامپیوتر

class CPU():
    def __init__(self, model):
        self.model = model

    def process(self):
        return f"CPU {self.model} is processing data ..."
    
class RAM():
    def __init__(self, size):
        self.size = size

    def load(self):
        return f"Loading data into {self.size}GB RAM ..."
    
class Computer():
    def __init__(self, cpu_model, ram_aize):
        self.cpu = CPU(cpu_model)
        self.ram = RAM(ram_aize)

    def start(self):
        print(self.cpu.process)
        print(self.ram.load)
        print("computer Started Successfully !!!")


my_pc = Computer("Core i7", 16)
my_pc.start()
```

## Multiple Inheritance (وراثت چندگانه)

- در وراثت چندگانه یک کلاس می‌تواند از چند والد ارث میبرد

- Child → ParentA + ParentB



```python
# مثال 

class A:
    def show(self):
        print("A")

class B:
    def show(self):
        print("B")

class C(A, B):
    pass

c = C()
c.show()  # => A

# چون ترتیب وراثت C(A, B) است.

print(C.mro())  
# یا
print(C.__mro__)
```

 ### پایتون چگونه تصمیم می‌گیرد؟

پایتون از یک قانون استفاده می‌کند به نام:
⭐ C3 Linearization → MRO

برای اینکه پایتون به صورت یکنواخت و بدون تناقض بداند:

    اول از کدام کلاس استفاده کند
    اگر در چند کلاس متد بود، کدام اولویت دارد
    اگر والدها مشترک بودند، از تکرار جلوگیری شود

```python
#  مثال کلاسیک Diamond Problem
# اگر کلاس A یک متد داشته باشد و D آن را صدا بزند…

class A:
    def process(self):
        print("A")

class B(A):
    def process(self):
        print("B")

class C(A):
    def process(self):
        print("C")

class D(B, C):
    pass

d = D()
d.process()
# خروجی می شود B




print(D.mro())
```

پس پایتون می‌گوید:

    ببین D چیزی دارد؟
    اگر نه → برو سراغ B
    اگر نه → برو C
    اگر نه → برو A
