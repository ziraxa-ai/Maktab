## list Metods

.append()  <=> افزودن یک عنصر به انتهای لیست

.extend()  <=>افزودن چند عنصر (یک لیست دیگر)

.insert()  <=> افزودن یک عنصر در یک اندیس مشخص 

.remove()  <=> حذف اولین مقدار در لیست 

.pop()  <=> حذف و برگردادن عنصر

.clear()  <=> حذف تمامی عناصر 

.index()  <=> پیدا کردن اندیس یک عنصر

.count()  <=>  شمارش تکرار یک عنصر

.sort()   <=>  مرتب کردن یک لیست <=>(reverse <=> مشخص کردن جهت و یا ترتیب)

.reverse()  <=> معکوس کردن 

.len()  <=> تعداد عناصر را بهت میگه 

.sum()  <=>  جمع داده ها 

.min(), max()  <=>  کمترین و بیشترین مقدار
