### Case Methods (متد های تغییر ظاهر حروف )

.capitalize() <=> اولین حرف را بزرگ و بقیه را کوچک می کند 

.casefold() <=> (مناسب کارکردن با تمام زبان ها است)تبدیل به حروف کوچک

.lower() <=> تمام حروف را کوچک می کند 

.upper() <=> تمام حروف را بزرگ میکند 

.swapcase()  <=>  حروف بزرگ و کوچک را جابجا می کند 

.title()  <=>  ابتدای هر کلمه را بزرگ میکند 



```python
massage = "TahA SAaDat"

print("Massage_capitalize :", massage.capitalize())
print("Massage_casefold :", massage.casefold())
print("Massage_lower:", massage.lower())
print("Massage_upper:", massage.upper())
print("Massage_swapcase :", massage.swapcase())
print("Massage_title:", massage.title())
```

### Check / Boolean Methods (متدهای بررسی محتوا)
.isalnum() <=> آیا فقط  حرف و عدد ؟

.isalpha() <=> آیا فقط حروف است ؟

.isascii() <=> فقط کاراکتر هایASCll

.isdecimal() <=> فقط ارقام ده دهی

.isdigit() <=> فقط رقم ؟

.isnumeric() <=> فقط عدد و یونیک هستند؟

.isidentifier() <=> نام برای متغیر مناسب است ؟

.islower() <=> ایا همه حروف کوچک است ؟

.isupper() <=> آیا همه حروف بزرگ است ؟

.isspace() <=> آیا تمام رشته فاصله (space, tab, newline)؟ هست 

.istitle() <=>بررسی می‌کند آیا رشته به فرم Title Case است یا نه.

.isprintable() <=> ایا قابل پرینت است 

```python
print("massage_isalnum : ", massage.isalnum())
print("massage_isalpha : ", massage.isalpha())
print("massage_isascii : ", massage.isascii())
print("massage_isdecimal : ", massage.isdecimal())
print("massage_isdigit : ", massage.isdigit())
print("massage_isnumeric : ", massage.isnumeric())
print("massage_isidentifier : ", massage.isidentifier())
print("massage_islower : ", massage.islower())
print("massage_isupper : ", massage.isupper())
print("massage_isspace : ", massage.isspace())
print("massage_istitle : ", massage.istitle())
print("massage_isprintable : ", massage.isprintable())
print("-------------title case------------")
print("Hello World".istitle())      # True
print("Hello world".istitle())      # False
print("HELLO World".istitle())      # False
print("Python Is Fun".istitle())    # True
```

## جدا کردن رشته به لیست

.split()  <=> جدا کردن رشته به لیست

.rsplit()  <=> جدا کردن از راست

.splitlines()  <=> جدا کردن بر اساس خط جدید

.partition(sep)  <=> تقسیم به (قبل, جداکننده, بعد)

.rpartition(sep)  <=> تقسیم از راست

```python
massage = "TahA SAaDat"

print("massage _ split : ", massage.split())
print("massage _ rsplit : ", massage.rsplit())
print("massage _ splitlines : ", massage.splitlines())
print("massage _ partition : ", massage.partition(massage[:1]))
print("massage _ rpartition : ", massage.rpartition(massage[:1]))
```

## متد های حذف فاصله و کاراکتر

.strip() <=> حذف از ابتدا و انتها

.lstrip() <=> حذف از چپ

.rstrip() <=> حذف از راست

.removeprefix(prefix) <=>اگر رشته با prefix شروع شود، آن را حذف می‌کند

.removesuffix(suffix) <=> حذف پسوند مشخص

```python
massage = "TahA SAaDat"

print("massage_strip() ", massage.strip())
print("massage_lstrip() ", massage.lstrip())
print("massage_rstrip() ", massage.rstrip())

# ----------------------------------
text = "https://example.com"
print(text.removeprefix("https://"))
# ----------------------------------
filename = "report.pdf"
print(filename.removesuffix(".pdf"))
```
