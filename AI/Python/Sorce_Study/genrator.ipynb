## ساخت تابع جنرتور با yield


```python
# تابع معمولی
def normal_function():
    return 1
# تابع جنریتور
def my_generator():
    yield 1
    yield 2
    yield 3

g = my_generator()

print(next(g))
print(next(g))
print(next(g))

```

```python
# EX 01 

def count_up_to(n):
    i = 1
    while i <= n:
        yield i
        i += 1

for number in count_up_to(10):
    print(number)
```

```python
#  به عنوان تحلیلگر داده روش قدیمی اینجوری بود که 

with open("data.txt") as f:
    lines = f.readlines()

#  کل رم را استفاده میکرد 

def read_lines(file_path):
    with open(file_path) as f:
        for line in f:
            yield line

for line in read_lines("data.txt"):
    print(line)
```

```python
def even_numbers(n):
    for i in range(n):
        if i % 2 == 0:
            yield i
        else:
            continue


for i in even_numbers(10):
    print(i)
```

```python
# EX 02
# می خواهیم یک برنامه بریزیم که تاس های مسابقه تخته فردا رو در یک فایل ذخیره کند
import random
import pickle

data = []

for i in range(10_000):
    this_one = (random.randint(1, 6), random.randint(1, 6))
    data.append(this_one)

taas_out = open("mosabeghe.taas", "wb")
pickle.dump(data, taas_out)        # به کمک این کتابخانه ما فایل هایی میسازیم و می توانیم از اان جای دیگر ران و استفاده کنیم 
```

```python
#EX 02.2
# حالا می خواهیم فایل ها را بخوانیم 

def get_taas():
    with open("mosabeghe.taas", "rb") as data_file:
        data = pickle.load(data_file)

    for taas in data:
        yield taas


# EX 02.3
#  GAME

dast = 0

for taas in get_taas():
    print(taas)
    if dast > 5:
        break
    else:
        dast += 1
```
