### for

### while

- حلقه تکراری که در طول انجام کار شرط را چک می کند و وارد حلقه میشود 
- معنی وایل می شود تا هنگامی که

```python
a = 0

while a <= 5:
    print(f"a is {a}")
    a += 1 
```

### دستور های حلقه ها 

- برای زمان هایی که می خواهیم یک شرط را برقرار کنیم و یا روی برنامه یک کنترل ویژه داشته باشیم 
- pass <=> کاری نمی کند فقط برای زمانی گذاشته می شود که جای خالی را پر کند و ادامه برنامه را ران کند 
- break <=> برای خارج شدن از حلقه است
- continue <=> برای زمانی است که می خواهد حلقه به ابتدا برگردد

```python
num = 0

while num <= 20:
    num += 1
    if num / 2 == 0:
        print(f"number is : {num }")
    else :
        continue
    if num == 18 :
        break

```
