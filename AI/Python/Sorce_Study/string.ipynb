- دسترسی به یک ایندکس <=> [شماره ایندکس]

```python
massage = "hello friends , to day is good day's"
print(massage[1])
print(massage[0:5])
print(massage[0 : 5 : 2])
print(massage[-1])
print(massage[0 : -1])
print(massage[::2]) # کلش رو بنویس دوتا دوتا
print(massage[::-1]) # همش رو می خوام ولی از آخر به اول
print
```

```python
name = "taha"
# روش اشتباه 

# name[0] = "T"


# روش تنبلی

print (name)

name = "Taha"
print (name)
# روش دوم 

name = "T" + name[1:]
print (name)


# مثال محاسبات 
print((name + "  ") * 3)
```

## ترکیب رشته ها (STRING Interpolsttion)

### format string



```python
# روش تنبلی
name = "taha"
family_name = "saadat"
print(name + "  "+"family :" + family_name)


# (روش دوم پیشنهاد نمیشود 

# print("hello %s welcome :) "%(name, family_name))

# روش سوم (پیشنهادی ) 
print("hello {} {} welcome".format(name, family_name))
print("hello {1} {0} welcome".format(name, family_name))
print("hello {family} {name} welcome".format(name = name, family = family_name))

# مشخص کردن دقت 
age = 22.2432432
print("hello {name} {family} welcome {name} have age : {sen: 1.1f}".format(name = name, family = family_name, sen = age))
# سنم یک مقدار عجیب قریب بود خواستم بهش مقدارم بدم با یک رقم اعشار چاپ شود 


# روش چهارم 
print(f"hello {name}, i donow your family {family_name} you are {age : 1.1f} years old")
```
