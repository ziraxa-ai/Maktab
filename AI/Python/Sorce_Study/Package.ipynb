## Package (Library)

- مجموعه کتابخانه ای که برای استفاده در یک شاخه خاص از برنامه نویسی ان را ایمپورت میکنیم و با دانش مورد نیاز مار میکنیم

- برای کار یا سرچ کردن کتابخانه های خاص در سایت (pypi) سرچ میکنیم

- برای دانلود کتابخانه هایی ک روی سیستممون نداریم از 

- => pip install pip_name

## package

- کمی بزرگتر از ماژل است و برای داشتن ان ما نیاز داریم یک فولدر  به نام پکیج داشته باشیم 

- نام فوولدر پکیج است 

- درو آن باید حتما یک فایل داشته باشیم با عنوان:(__init.py__)

- نیازی نیست پر باشد اما حتما باید باشد  

### (__name__)

- اسم چیزی که داره اجرا میشود ، درنتیجه برنامه می فهمد که آیا من یک پکیجم که در یک برنامه ران میشم یا خودم مستقیم ران میشوم


```python
# یکی از کاربرد های بسیار رایج آن این است که 

if __name__ == "__main__":
    print("You should use me as a package !!!")


# زمانی که یکی فایل ماژل را ران کند خروجی جمله بالا است و یا درخواستی که به شکل پایه برای آن در نظر میگیریم اما اگر در فایل برنامه ران شود براساس برنامه ای که نوشتیم 
#  مثلا در برنامه ای که برای ساخت نمودار نوشتیم می توانیم شرط بزاریم اگر کسی فایل را ران کرد اطلاعاتی بگیرد و درخواستش  را اننجام دهد.
```
